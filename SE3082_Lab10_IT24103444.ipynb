## Exercise 1(a) – GPU Timing and Synchronization

```python
%%writefile ex1_timed.cu
#include <cstdio>
#include <cstdlib>
#include <sys/time.h>

#define N 512

__global__ void add(int *a, int *b, int *c) {
    c[blockIdx.x] = a[blockIdx.x] + b[blockIdx.x];
}

void random_ints(int *x, int size) {
    for (int i = 0; i < size; i++)
        x[i] = rand() % 100;
}

double cpuSecond() {
    struct timeval tp;
    gettimeofday(&tp, NULL);
    return (double)tp.tv_sec + (double)tp.tv_usec * 1.e-6;
}

int main(void) {
    int *a, *b, *c;
    int *d_a, *d_b, *d_c;
    int size = N * sizeof(int);
    double dStart, dElaps;

    cudaMalloc((void **)&d_a, size);
    cudaMalloc((void **)&d_b, size);
    cudaMalloc((void **)&d_c, size);

    a = (int *)malloc(size);
    random_ints(a, N);

    b = (int *)malloc(size);
    random_ints(b, N);

    c = (int *)malloc(size);

    cudaMemcpy(d_a, a, size, cudaMemcpyHostToDevice);
    cudaMemcpy(d_b, b, size, cudaMemcpyHostToDevice);

    dStart = cpuSecond();

    add<<<N, 1>>>(d_a, d_b, d_c);

    cudaDeviceSynchronize();

    dElaps = cpuSecond() - dStart;

    cudaMemcpy(c, d_c, size, cudaMemcpyDeviceToHost);

    for (int r = 0; r < 5; r++)
        printf("%d + %d = %d\n", a[r], b[r], c[r]);

    int errors = 0;

    for (int r = 0; r < N; r++) {
        if (c[r] != a[r] + b[r])
            errors++;
    }

    printf("Verification: %d mismatches out of %d\n", errors, N);
    printf("Time taken - %.7f\n", dElaps);

    free(a);
    free(b);
    free(c);

    cudaFree(d_a);
    cudaFree(d_b);
    cudaFree(d_c);

    return 0;
}
```

```python
%%writefile ex5_vecmul_timed.cu
#include <cstdio>
#include <cstdlib>
#include <sys/time.h>
#include <cuda_runtime.h>

#define N 10000000
#define THREADS_PER_BLOCK 512

__global__ void vecMul(const int *a, const int *b, int *c, int n) {
    int i = blockIdx.x * blockDim.x + threadIdx.x;

    if (i < n)
        c[i] = a[i] * b[i];
}

double cpuSecond() {
    struct timeval tp;
    gettimeofday(&tp, NULL);
    return (double)tp.tv_sec + (double)tp.tv_usec * 1.e-6;
}

int main() {
    size_t size = (size_t)N * sizeof(int);

    int *a = (int*)malloc(size);
    int *b = (int*)malloc(size);
    int *c = (int*)malloc(size);

    if (!a || !b || !c) {
        printf("Host memory allocation failed\n");
        return 1;
    }

    srand(0);

    for (int i = 0; i < N; i++)
        a[i] = rand() % 100;

    for (int i = 0; i < N; i++)
        b[i] = rand() % 100;

    int *d_a, *d_b, *d_c;

    cudaMalloc((void**)&d_a, size);
    cudaMalloc((void**)&d_b, size);
    cudaMalloc((void**)&d_c, size);

    cudaMemcpy(d_a, a, size, cudaMemcpyHostToDevice);
    cudaMemcpy(d_b, b, size, cudaMemcpyHostToDevice);

    int blocks = (N + THREADS_PER_BLOCK - 1) / THREADS_PER_BLOCK;

    double dStart = cpuSecond();

    vecMul<<<blocks, THREADS_PER_BLOCK>>>(d_a, d_b, d_c, N);
    cudaError_t err = cudaDeviceSynchronize();

    double dElaps = cpuSecond() - dStart;

    if (err != cudaSuccess) {
        printf("CUDA error: %s\n", cudaGetErrorString(err));
        return 1;
    }

    cudaMemcpy(c, d_c, size, cudaMemcpyDeviceToHost);

    int mismatches = 0;

    for (int i = 0; i < N; i++) {
        if (c[i] != a[i] * b[i])
            mismatches++;
    }

    printf("Blocks: %d, Threads per block: %d\n",
           blocks, THREADS_PER_BLOCK);
    printf("Verification: %d mismatches out of %d\n", mismatches, N);
    printf("Time taken - %.7f\n", dElaps);

    cudaFree(d_a);
    cudaFree(d_b);
    cudaFree(d_c);

    free(a);
    free(b);
    free(c);

    return 0;
}
```

## Exercise 1(c) – Kernel Time vs Total Execution Time

```python
%%writefile ex1_copy_vs_kernel.cu
#include <cstdio>
#include <cstdlib>
#include <sys/time.h>
#include <cuda_runtime.h>

#define N 100000000
#define THREADS_PER_BLOCK 512

#define CHECK(call) do { \
    cudaError_t err = (call); \
    if (err != cudaSuccess) { \
        printf("CUDA error: %s\n", cudaGetErrorString(err)); \
        exit(1); \
    } \
} while (0)

__global__ void vecMul(const int *a, const int *b, int *c, int n) {
    int i = blockIdx.x * blockDim.x + threadIdx.x;
    if (i < n)
        c[i] = a[i] * b[i];
}

double cpuSecond() {
    struct timeval tp;
    gettimeofday(&tp, NULL);
    return (double)tp.tv_sec + (double)tp.tv_usec * 1.e-6;
}

int main() {
    size_t size = (size_t)N * sizeof(int);

    int *a = (int*)malloc(size);
    int *b = (int*)malloc(size);
    int *c = (int*)malloc(size);

    if (!a || !b || !c) {
        printf("Host memory allocation failed\n");
        return 1;
    }

    srand(0);
    for (int i = 0; i < N; i++) a[i] = rand() % 100;
    for (int i = 0; i < N; i++) b[i] = rand() % 100;

    int *d_a, *d_b, *d_c;
    CHECK(cudaMalloc((void**)&d_a, size));
    CHECK(cudaMalloc((void**)&d_b, size));
    CHECK(cudaMalloc((void**)&d_c, size));

    int blocks = (N + THREADS_PER_BLOCK - 1) / THREADS_PER_BLOCK;

    // Whole sequence starts before copying inputs
    double totalStart = cpuSecond();

    CHECK(cudaMemcpy(d_a, a, size, cudaMemcpyHostToDevice));
    CHECK(cudaMemcpy(d_b, b, size, cudaMemcpyHostToDevice));

    // Kernel-only timer
    double kernelStart = cpuSecond();

    vecMul<<<blocks, THREADS_PER_BLOCK>>>(d_a, d_b, d_c, N);

    CHECK(cudaGetLastError());
    CHECK(cudaDeviceSynchronize());

    double kernelTime = cpuSecond() - kernelStart;

    CHECK(cudaMemcpy(c, d_c, size, cudaMemcpyDeviceToHost));

    double totalTime = cpuSecond() - totalStart;

    size_t mismatches = 0;
    for (int i = 0; i < N; i++) {
        if (c[i] != a[i] * b[i])
            mismatches++;
    }

    printf("Verification: %zu mismatches out of %d\n", mismatches, N);
    printf("Kernel-only time: %.7f seconds\n", kernelTime);
    printf("Whole-sequence time: %.7f seconds\n", totalTime);

    CHECK(cudaFree(d_a));
    CHECK(cudaFree(d_b));
    CHECK(cudaFree(d_c));

    free(a);
    free(b);
    free(c);

    return 0;
}
```

## Exercise 2 – CUDA Error Checking

```python
%%writefile ex2_check.cu
#include <stdio.h>
#include <stdlib.h>
#include <sys/time.h>
#include <cuda_runtime.h>

#define N 512

#define CHECK(call) do { \
    const cudaError_t error = (call); \
    if (error != cudaSuccess) { \
        printf("Error: %s:%d, ", __FILE__, __LINE__); \
        printf("code:%d, reason: %s\n", error, cudaGetErrorString(error)); \
        exit(1); \
    } \
} while (0)

__global__ void add(int *a, int *b, int *c) {
    c[blockIdx.x] = a[blockIdx.x] + b[blockIdx.x];
}

double cpuSecond() {
    struct timeval tp;
    gettimeofday(&tp, NULL);
    return (double)tp.tv_sec + (double)tp.tv_usec * 1.e-6;
}

int main() {
    int *a, *b, *c;
    int *d_a, *d_b, *d_c;
    size_t size = N * sizeof(int);

    a = (int*)malloc(size);
    b = (int*)malloc(size);
    c = (int*)malloc(size);

    for (int i = 0; i < N; i++) {
        a[i] = rand() % 100;
        b[i] = rand() % 100;
    }

    CHECK(cudaMalloc((void**)&d_a, size));
    CHECK(cudaMalloc((void**)&d_b, size));
    CHECK(cudaMalloc((void**)&d_c, size));

    CHECK(cudaMemcpy(d_a, a, size, cudaMemcpyHostToDevice));
    CHECK(cudaMemcpy(d_b, b, size, cudaMemcpyHostToDevice));

    double dStart = cpuSecond();

    add<<<N, 1>>>(d_a, d_b, d_c);

    CHECK(cudaGetLastError());
    CHECK(cudaDeviceSynchronize());

    double dElaps = cpuSecond() - dStart;

    CHECK(cudaMemcpy(c, d_c, size, cudaMemcpyDeviceToHost));

    for (int i = 0; i < 5; i++)
        printf("%d + %d = %d\n", a[i], b[i], c[i]);

    int mismatches = 0;
    for (int i = 0; i < N; i++)
        if (c[i] != a[i] + b[i])
            mismatches++;

    printf("Verification: %d mismatches out of %d\n", mismatches, N);
    printf("Time taken - %.7f\n", dElaps);

    CHECK(cudaFree(d_a));
    CHECK(cudaFree(d_b));
    CHECK(cudaFree(d_c));

    free(a);
    free(b);
    free(c);

    return 0;
}
```

## Exercise 3 – 2D Picture Kernel

```python
%%writefile ex3_picture.cu
#include <cstdio>
#include <cstdlib>
#include <sys/time.h>
#include <cuda_runtime.h>

#define WIDTH 800
#define HEIGHT 600

#define CHECK(call) do { \
    cudaError_t err = (call); \
    if (err != cudaSuccess) { \
        printf("CUDA Error: %s\n", cudaGetErrorString(err)); \
        exit(1); \
    } \
} while (0)

__global__ void PictureKernel(
    unsigned char *d_Pin,
    unsigned char *d_Pout,
    int width,
    int height
) {
    int Col = blockIdx.x * blockDim.x + threadIdx.x;
    int Row = blockIdx.y * blockDim.y + threadIdx.y;

    if (Row < height && Col < width) {
        int index = Row * width + Col;
        int value = 2 * (int)d_Pin[index];

        d_Pout[index] = (value > 255) ? 255 : value;
    }
}

double cpuSecond() {
    struct timeval tp;
    gettimeofday(&tp, NULL);
    return (double)tp.tv_sec + (double)tp.tv_usec * 1.e-6;
}

int main() {
    const int N = WIDTH * HEIGHT;
    const size_t size = (size_t)N * sizeof(unsigned char);

    unsigned char *h_Pin = (unsigned char*)malloc(size);
    unsigned char *h_Pout = (unsigned char*)malloc(size);

    if (!h_Pin || !h_Pout) {
        printf("Host memory allocation failed\n");
        return 1;
    }

    for (int i = 0; i < N; i++)
        h_Pin[i] = rand() % 256;

    unsigned char *d_Pin, *d_Pout;

    CHECK(cudaMalloc((void**)&d_Pin, size));
    CHECK(cudaMalloc((void**)&d_Pout, size));

    CHECK(cudaMemcpy(d_Pin, h_Pin, size, cudaMemcpyHostToDevice));

    dim3 threadsPerBlock(16, 16);
    dim3 numBlocks(
        (WIDTH + 15) / 16,
        (HEIGHT + 15) / 16
    );

    double dStart = cpuSecond();

    PictureKernel<<<numBlocks, threadsPerBlock>>>(
        d_Pin, d_Pout, WIDTH, HEIGHT
    );

    CHECK(cudaGetLastError());
    CHECK(cudaDeviceSynchronize());

    double dElaps = cpuSecond() - dStart;

    CHECK(cudaMemcpy(h_Pout, d_Pout, size, cudaMemcpyDeviceToHost));

    int mismatches = 0;
    int cappedPixels = 0;

    for (int i = 0; i < N; i++) {
        int expected = 2 * (int)h_Pin[i];

        if (expected > 255) {
            expected = 255;
            cappedPixels++;
        }

        if (h_Pout[i] != expected)
            mismatches++;
    }

    printf("Image size: %d x %d\n", WIDTH, HEIGHT);
    printf("Grid: %u x %u blocks\n", numBlocks.x, numBlocks.y);
    printf("Threads per block: 16 x 16\n");

    printf("\nOriginal pixels (first 50 x 50):\n");

    for (int row = 0; row < 50; row++) {
        for (int col = 0; col < 50; col++)
            printf("%3d ", (int)h_Pin[row * WIDTH + col]);

        printf("\n");
    }

    printf("\nUpdated pixels (first 50 x 50):\n");

    for (int row = 0; row < 50; row++) {
        for (int col = 0; col < 50; col++)
            printf("%3d ", (int)h_Pout[row * WIDTH + col]);

        printf("\n");
    }

    printf("\nVerification: %d mismatches out of %d\n", mismatches, N);
    printf("Capped pixels: %d\n", cappedPixels);
    printf("Time taken - %.7f\n", dElaps);

    CHECK(cudaFree(d_Pin));
    CHECK(cudaFree(d_Pout));

    free(h_Pin);
    free(h_Pout);

    return 0;
}
```

## Exercise 4 – 3D Array Multiplication

```python
%%writefile ex4_volume.cu
#include <cstdio>
#include <cstdlib>
#include <sys/time.h>
#include <cuda_runtime.h>

#define NX 500
#define NY 500
#define NZ 500

#define CHECK(call) do { \
    cudaError_t err = (call); \
    if (err != cudaSuccess) { \
        printf("CUDA Error: %s\n", cudaGetErrorString(err)); \
        exit(1); \
    } \
} while (0)

__global__ void multiply3D(
    const int *a, const int *b, int *c,
    int nx, int ny, int nz
) {
    int x = blockIdx.x * blockDim.x + threadIdx.x;
    int y = blockIdx.y * blockDim.y + threadIdx.y;
    int z = blockIdx.z * blockDim.z + threadIdx.z;

    if (x < nx && y < ny && z < nz) {
        size_t index = ((size_t)z * ny + y) * nx + x;
        c[index] = a[index] * b[index];
    }
}

double cpuSecond() {
    struct timeval tp;
    gettimeofday(&tp, NULL);
    return (double)tp.tv_sec + (double)tp.tv_usec * 1.e-6;
}

int main() {
    size_t count = (size_t)NX * NY * NZ;
    size_t size = count * sizeof(int);

    int *a = (int*)malloc(size);
    int *b = (int*)malloc(size);
    int *c = (int*)malloc(size);

    if (!a || !b || !c) {
        printf("Host memory allocation failed\n");
        return 1;
    }

    for (size_t i = 0; i < count; i++)
        a[i] = rand() % 100;

    for (size_t i = 0; i < count; i++)
        b[i] = rand() % 100;

    int *d_a, *d_b, *d_c;

    CHECK(cudaMalloc((void**)&d_a, size));
    CHECK(cudaMalloc((void**)&d_b, size));
    CHECK(cudaMalloc((void**)&d_c, size));

    CHECK(cudaMemcpy(d_a, a, size, cudaMemcpyHostToDevice));
    CHECK(cudaMemcpy(d_b, b, size, cudaMemcpyHostToDevice));

    dim3 threadsPerBlock(8, 8, 8);

    dim3 numBlocks(
        (NX + 7) / 8,
        (NY + 7) / 8,
        (NZ + 7) / 8
    );

    double dStart = cpuSecond();

    multiply3D<<<numBlocks, threadsPerBlock>>>(
        d_a, d_b, d_c, NX, NY, NZ
    );

    CHECK(cudaGetLastError());
    CHECK(cudaDeviceSynchronize());

    double dElaps = cpuSecond() - dStart;

    CHECK(cudaMemcpy(c, d_c, size, cudaMemcpyDeviceToHost));

    size_t mismatches = 0;

    for (size_t i = 0; i < count; i++) {
        if (c[i] != a[i] * b[i])
            mismatches++;
    }

    printf("3D dimensions: %d x %d x %d\n", NX, NY, NZ);
    printf("Grid: %u x %u x %u blocks\n",
           numBlocks.x, numBlocks.y, numBlocks.z);
    printf("Threads per block: 8 x 8 x 8\n");

    printf("\nLast 1000 elements:\n");

    for (size_t i = count - 1000; i < count; i++) {
        size_t z = i / ((size_t)NX * NY);
        size_t y = (i / NX) % NY;
        size_t x = i % NX;

        printf("C[%zu][%zu][%zu] = %d x %d = %d\n",
               z, y, x, a[i], b[i], c[i]);
    }

    printf("\nVerification: %zu mismatches out of %zu\n",
           mismatches, count);
    printf("Time taken - %.7f\n", dElaps);

    CHECK(cudaFree(d_a));
    CHECK(cudaFree(d_b));
    CHECK(cudaFree(d_c));

    free(a);
    free(b);
    free(c);

    return 0;
}
```
